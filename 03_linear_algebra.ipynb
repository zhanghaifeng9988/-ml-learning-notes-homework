```python
# 编写1个函数，来进行向量乘法
def vector_vector_multiplication(u,v):
    assert  u.shape[0] == v.shape[0]  # 检查向量形状 # 断言条件成立，如果不成立，程序立刻崩溃报错

    n = u.shape[0] 
    result = 0.0 
    for i in range(n):
        result = result + u[i] * v[i]
    return result
```

```python
vector_vector_multiplication(u,v)
```

```python
u.dot(v)  #numpy中的点积乘法
```

## matirx_vector_Multiplication  矩阵向量乘法

```python
def matrix_vector_multiplication(U,v):
    assert U.shape[1] == v.shape[0]  # 判断矩阵的列数，与向量元素是否相等，表示矩阵和向量相乘，即是向量乘以矩阵的每一行
    num_rows = U.shape[0] 
    result = np.zeros(num_rows)
    for i in range(num_rows):
       result[i] = vector_vector_multiplication(U[i],v) 
    return result
```

```python
matrix_vector_multiplication(U,v)
```

```python
U.dot(v)   #numpy中的点积乘法 ,结果是一样的。
```

## 矩阵乘法 # 关键是，两个矩阵中，1个的行的shape的值，与1个的列shape的值，是要一样的

```python
V = np.array([[1,1,2],[0,0.5,1],[0,2,1],[2,1,0]])
```

```python
V
```

```python
def matrix_matrix_multiplication(U,V):
    assert U.shape[1] == V.shape[0] # 判断矩阵的行列是否相等，表示矩阵相乘即是 列乘以行
    num_rows = U.shape[0] 
    num_cols = V.shape[1] 
    result = np.zeros((num_rows,num_cols)) # 3行3列的矩阵

    for i in range(num_cols):
        vi = V[:,i]
        Uvi = matirx_vector_multiplication(U,vi)
        result[:,i] = Uvi
    return result
```

# matrix  inverse  矩阵逆：一个矩阵的逆矩阵，当它与原矩阵相乘时，结果为***单位矩阵***。
***只有方阵才有逆矩阵***

```python
Vs = V[[0,1,2]]  #把V矩阵中的1，2，3行取出来，就是3*3的方阵
Vs
```

```python
Vs_inv = np.linalg.inv(Vs)  # 求Vs的逆矩阵 
Vs_inv
```

```python
Vs.dot(Vs_inv)
```
